```python
import statistics as stats

def compute(data):
    # Calculate Mean
    mean = stats.mean(data)

    # Calculate Median
    median = stats.median(data)

    # Calculate Mode
    try:
        mode = stats.mode(data)
    except stats.StatisticsError: 
        mode = "No Unique Mode"

    # Calculate Variance
    variance = stats.variance(data)

    # Calculate Standard Deviation
    std_dev = stats.stdev(data)

    return mean, median, mode, variance, std_dev

data = [10, 20, 20, 30, 40, 40, 40, 50, 60, 70, 80, 80, 100]
mean, median, mode, variance, std_dev = compute(data)

table = []
for x in data:
    x_minus_mean = x - mean
    x_minus_mean_squared = x_minus_mean ** 2
    table.append([x, mean, x_minus_mean, x_minus_mean_squared])

print("\n\t\t\t--------------------- Program to Compute the Central Tendency Measures ---------------------------\n")
print("\t--------------------------- Calculation Table ---------------------------------------")
print('\t' + '-' * 45)
print("\t| {:<3} | {:<6} | {:<6} | {:<7} |".format('x', 'Mean', 'x-Mean', '(x-Mean)²'))
print('\t' + '-' * 45)

for row in table:
    rounded_value = round(row[3], 1)
    print("\t| {:<3} | {:<6.2f} | {:<6.2f} | {:<8.1f} |".format(row[0], row[1], row[2], rounded_value)) 
print('\t' + '-' * 45)

print("\n\t" + '-' * 15 + " Central Tendency Measures " + '-' * 18)
print("\n\t" + '-' * 60)
print("\t| {:<10} | {:<10} | {:<10} | {:<10} | {:<3} |".format('Mean', 'Median', 'Mode', 'Variance', 'Std_dev'))
print("\t" + '-' * 60)
print("\t| {:<10.2f} | {:<10.2f} | {:<10} | {:<10.2f} | {:<7.2f} |".format(mean, median, mode, variance, std_dev))
print("\t" + '-' * 60)
```
