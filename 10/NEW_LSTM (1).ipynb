```python
# Importing necessary libraries
import numpy as np
import tensorflow as tf
from tensorflow.keras.models import Sequential
from tensorflow.keras.layers import LSTM, Dense

# Sample data (sine wave example)
def generate_data(seq_length=100, n_samples=1000):
    X = np.linspace(0, 50, seq_length * n_samples)
    y = np.sin(X)
    X = X.reshape(n_samples, seq_length, 1)
    y = y.reshape(n_samples, seq_length, 1)
    return X, y

# Generate sample data
X, y = generate_data()

# LSTM model
model = Sequential()

# Adding LSTM layer (return_sequences=True because we want to output sequences)
model.add(LSTM(50, activation='relu', return_sequences=True, input_shape=(100, 1)))

# Adding Dense layer for output
model.add(Dense(1))

# Compiling the model
model.compile(optimizer='adam', loss='mean_squared_error')

# Training the model
model.fit(X, y, epochs=10, batch_size=32)

# Summary of the model
model.summary()

# Making predictions (using the same input for simplicity)
predictions = model.predict(X)
print(predictions[:5])
```
